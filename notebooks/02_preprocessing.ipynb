```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/googleplaystore.csv")
df = df.drop_duplicates()
```

```python
df.info()
```

```python
df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")
```

```python
df["Reviews"].dtype
```

```python
df["Installs"] = (
    df["Installs"]
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)

df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")
```

```python
df["Installs"].head()
```

```python
df["Price"] = (
    df["Price"]
    .str.replace("$", "", regex=False)
)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
```

```python
df["Price"].head()
```

```python
df["Size"].unique()
```

```python
df["Size"] = df["Size"].replace("Varies with device", np.nan)
```

```python
def convert_size(size):
    if pd.isna(size):
        return np.nan

    if size.endswith("M"):
        return float(size[:-1])

    if size.endswith("k"):
        return float(size[:-1]) / 1024

    return np.nan
```

```python
df["Size"] = df["Size"].apply(convert_size)
```

```python
df.isnull().sum()
```

```python
df["Rating"] = df["Rating"].fillna(df["Rating"].median())

df["Size"] = df["Size"].fillna(df["Size"].median())
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df = df.dropna(subset=["Reviews", "Installs", "Type", "Price", "Content Rating"])
```

```python
df = df.drop(columns=["Current Ver", "Android Ver"])
```

```python
df.info()
```

```python
df.to_csv("../data/googleplaystore_clean.csv", index=False)
```
